```python
import dlt
from pyspark.sql import functions as F

@dlt.table(
    name="dlt_raw_sales",
    comment="Bronze layer: Raw retail transactions ingested automatically."
)
def dlt_raw_sales():
    # In a real project, this would be a streaming read using Auto Loader:
    # return spark.readStream.format("cloudFiles").option("cloudFiles.format", "json").load("/path")
    
    # For our framework baseline, we generate our mock dataset:
    mock_data = [
        (1001, "2026-07-17", 250.75),
        (1002, "2026-07-17", 19.99),
        (1003, "2026-07-17", 120.00),
        (None, "2026-07-17", 0.00) # Bad row to test quality later!
    ]
    
    df = spark.createDataFrame(mock_data, ["transaction_id", "date", "amount"])
    
    # Add an ingestion timestamp automatically
    return df.withColumn("ingested_at", F.current_timestamp())
```
