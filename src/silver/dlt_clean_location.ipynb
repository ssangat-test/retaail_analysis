```python
import dlt
from pyspark.sql import functions as F

@dlt.table(
    name="clean_location",
    comment="Silver: Cleaned store locations with standardized formatting."
)
def clean_location():
    return dlt.read("dlt_raw_location") \
        .withColumn("city", F.upper(F.col("city"))) \
        .withColumn("state", F.upper(F.col("state")))
```
