```python
import dlt
from pyspark.sql import functions as F

@dlt.table(
    name="clean_sales",
    comment="Silver: Cleaned sales transactions with proper types."
)
@dlt.expect_or_drop("valid_transaction", "transaction_id IS NOT NULL")
def clean_sales():
    # Injecting location IDs to match our location table for the join!
    return dlt.read("dlt_raw_sales") \
        .withColumn("date", F.col("date").cast("date")) \
        .withColumn("amount", F.col("amount").cast("double")) \
        .withColumn("location_id", F.when(F.col("transaction_id") == 1001, 1).otherwise(2))
```
