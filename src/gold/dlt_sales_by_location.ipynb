```python
import dlt
from pyspark.sql import functions as F

@dlt.table(
    name="total_sales_by_location",
    comment="Gold: Aggregated revenue per city for business dashboards."
)
def total_sales_by_location():
    df_sales = dlt.read("clean_sales")
    df_location = dlt.read("clean_location")
    
    # Enrich sales with geographic details and aggregate!
    return df_sales.join(df_location, "location_id", "inner") \
        .groupBy("city", "state") \
        .agg(F.sum("amount").alias("total_revenue"))
```
